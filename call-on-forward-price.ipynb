```python
import numpy as np
import pandas as pd
from scipy.stats import norm
```

```python
# Exercise 7.4
def black_call_forward(F, K, r, T, sigma):
    """  
    F = forward price
    K = strike price
    r = risk-free rate
    T = time to maturity
    sigma = volatility of the forward price

    returns black's price of call option
    """

    if sigma <= 0 or T<= 0:
        return max(0, F-K) * np.exp(-r*T)
    
    d1 = (np.log(F/K) + 0.5*sigma**2*T) / (sigma*np.sqrt(T))
    d2 = d1 - sigma * np.sqrt(T)
    call_price = np.exp(-r*T) * (F*norm.cdf(d1) - K*norm.cdf(d2))

    return call_price

def black_call_impliedvol(F, K, r, T, mkt_price, tolerance, max_iter):
    """  
    F = forward price
    K = strike price
    r = risk-free rate
    T = time to maturity
    mkt_price = market price of call option
    tolerance = threshold for convergence
    max_iter = maximum number of iteraction for bisection

    returns implied volatility from bisection method
    """

    # Range for volatilities
    lower_bound = 0.000001
    upper_bound = 5.0
    implied_vol = (lower_bound + upper_bound) / 2

    for i in range(max_iter):
        price = black_call_forward(F, K, r, T, implied_vol)

        if abs(price - mkt_price) < tolerance:
            return implied_vol
        
        if price < mkt_price:
            lower_bound = implied_vol
        else:
            upper_bound = implied_vol

    # In the event that tolerance is not achieved, return best guess:
    return implied_vol
```

```python
# Set parameters
F = 100
K = 105
r = 0.05
T = 1
mkt_price = 5
tolerance = 0.000001
max_iter = 1000

implied_vol = black_call_impliedvol(F, K, r, T, mkt_price, tolerance, max_iter)
print('Implied Volatility:', implied_vol)
```
