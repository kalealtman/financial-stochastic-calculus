```python
import numpy as np
import pandas as pd
```

```python
# Problem 2.1
# Sum for X = exp(t) 
def computeSum_21(T, N):

    deltaT = T / N

    computedSum = 0
    for i in range(1, N+1):
        ti = i * deltaT
        ti_prev = (i-1) * deltaT
        
        deltaX = np.exp(ti) - np.exp(ti_prev)
        computedSum = computedSum + deltaX * deltaX

    return computedSum
```

```python
# Example values to test
T = 1  # Total time
N = 100  # Number of intervals

result = computeSum_21(T, N)
print(result)
```

```python
# Problem 2.2
# Sum for X = t^3
def computeSum_22(T, N):

    deltaT = T / N

    computedSum = 0
    for i in range(1, N+1):
        ti = i * deltaT
        ti_prev = (i-1) * deltaT
        
        deltaX = ti**3 - ti_prev**3
        computedSum = computedSum + deltaX * deltaX

    return computedSum
```

```python
# Example values to test
T = 1  # Total time
N = 100  # Number of intervals

result = computeSum_22(T, N)
print(result)
```

```python
# Problem 2.3
# Sum for a Brownian motion
def computeSum_23(T, N):

    deltaT = T / N

    # Brownian Motion:
    B = np.zeros(N+1) # This sets up B(0) = 0
    for i in range(1, N+1):
        deltaB = np.random.normal(0, np.sqrt(deltaT))
        B[i] = B[i-1] + deltaB

    
    computedSum = 0
    for i in range(1, N+1):
        deltaB = B[i] - B[i-1]

        computedSum = computedSum + deltaB * deltaB

    return computedSum
```

```python
# Example values to test
T = 1  # Total time
N = 100  # Number of intervals

result = computeSum_23(T, N)
print(result)
```
